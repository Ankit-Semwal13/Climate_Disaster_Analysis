```python
#Visualization 1: Number of Disasters Over Time
#Business Question

#How has the frequency of natural disasters changed between 1970 and 2021?

import plotly.express as px

# Count disasters by year
yearly_disasters = (
    df.groupby("Year")
      .size()
      .reset_index(name="Number_of_Disasters")
)

fig = px.line(
    yearly_disasters,
    x="Year",
    y="Number_of_Disasters",
    title="Global Natural Disasters by Year (1970–2021)",
    markers=True
)

fig.update_layout(
    template="plotly_white",
    title_x=0.5,
    xaxis_title="Year",
    yaxis_title="Number of Disasters",
    height=600
)

fig.show()
```

```python
#Visualization 2: Top 10 Disaster Types
#Business Question

#Which natural disasters occur most frequently?

disaster_type = (
    df["Disaster Type"]
      .value_counts()
      .head(10)
      .reset_index()
)

disaster_type.columns = [
    "Disaster Type",
    "Count"
]

fig = px.bar(
    disaster_type,
    x="Disaster Type",
    y="Count",
    color="Count",
    text="Count",
    title="Top 10 Most Frequent Disaster Types"
)

fig.update_layout(
    template="plotly_white",
    title_x=0.5,
    height=600
)

fig.show()
```

```python
#Visualization 3: Disasters by Continent

continent = (
    df["Continent"]
      .value_counts()
      .reset_index()
)

continent.columns = [
    "Continent",
    "Count"
]

fig = px.pie(
    continent,
    values="Count",
    names="Continent",
    title="Distribution of Natural Disasters by Continent"
)

fig.update_layout(
    template="plotly_white",
    title_x=0.5,
    height=650
)

fig.show()
```

```python
#Visualization 4: Top 15 Countries
country = (
    df["Country"]
      .value_counts()
      .head(15)
      .reset_index()
)

country.columns = [
    "Country",
    "Count"
]

fig = px.bar(
    country,
    x="Count",
    y="Country",
    orientation="h",
    color="Count",
    title="Top 15 Countries with the Highest Number of Recorded Disasters"
)

fig.update_layout(
    template="plotly_white",
    title_x=0.5,
    height=700
)

fig.show()
```

```python
#Visualization 5: Hierarchy of Natural Disaster Types
treemap_data = (
    df.groupby(["Disaster Subgroup", "Disaster Type"])
      .size()
      .reset_index(name="Count")
)

fig = px.treemap(
    treemap_data,
    path=["Disaster Subgroup", "Disaster Type"],
    values="Count",
    title="Hierarchy of Natural Disaster Types"
)

fig.update_layout(
    title_x=0.5,
    template="plotly_white",
    height=700
)

fig.show()
```
