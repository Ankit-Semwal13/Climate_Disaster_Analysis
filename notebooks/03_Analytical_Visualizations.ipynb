# Analytical Question 1

## How has the frequency of natural disasters changed over time across different disaster types?

### Objective
This analysis investigates how the number of recorded natural disasters has changed from 1970 to 2021. It also examines whether all disaster types have followed similar trends or if certain types have become more frequent over time.

### Why is this important?
Understanding long-term trends helps identify changes in global disaster patterns and supports disaster preparedness and climate adaptation planning.

```python
import pandas as pd
import plotly.express as px

# Load dataset
df = pd.read_csv("../data/raw/emdat_disasters_1970_2021.csv")

print(df.shape)
```

```python
# Keep only the required columns
q1_df = df[
    [
        "Year",
        "Disaster Type"
    ]
].copy()

q1_df.head()
```

```python
q1_df.isnull().sum()
```

```python
year_disaster = (
    q1_df
    .groupby(["Year", "Disaster Type"])
    .size()
    .reset_index(name="Disaster Count")
)

year_disaster.head()
```

```python
fig = px.line(
    year_disaster,
    x="Year",
    y="Disaster Count",
    color="Disaster Type",
    markers=True,
    title="Trend of Natural Disaster Types (1970–2021)",
    labels={
        "Year": "Year",
        "Disaster Count": "Number of Recorded Disasters",
        "Disaster Type": "Disaster Type"
    }
)

fig.update_layout(
    template="plotly_white",
    title={
        "text": "Trend of Natural Disaster Types (1970–2021)",
        "x": 0.5,
        "font": {"size": 22}
    },
    xaxis=dict(showgrid=False),
    yaxis=dict(showgrid=True, gridcolor="lightgray"),
    legend_title="Disaster Type",
    height=650,
    hovermode="x unified"
)

fig.show()
```

## Interpretation

The visualization shows how the frequency of different natural disaster types has evolved between 1970 and 2021. It highlights whether specific disaster categories have increased more rapidly than others and reveals long-term changes in global disaster occurrence. These trends provide evidence for changing disaster patterns over time and help identify which disaster types require greater attention in future disaster management strategies.

# Analytical Question 2

## Which disaster types have caused the highest total number of deaths globally?

### Objective

This analysis compares different disaster types based on the total number of fatalities they have caused worldwide.

### Why is this important?

Understanding which disaster types are the deadliest helps governments and disaster management agencies prioritize preparedness, mitigation strategies, and resource allocation.

```python
# Select only required columns

q2_df = df[
    [
        "Disaster Type",
        "Total Deaths"
    ]
].copy()

q2_df.head()
```

```python
q2_df.isnull().sum()
```

```python
q2_df["Total Deaths"] = q2_df["Total Deaths"].fillna(0)
```

```python
death_summary = (
    q2_df
    .groupby("Disaster Type", as_index=False)
    ["Total Deaths"]
    .sum()
    .sort_values(
        by="Total Deaths",
        ascending=False
    )
)

death_summary
```

```python
fig = px.bar(
    death_summary,
    x="Total Deaths",
    y="Disaster Type",
    orientation="h",
    color="Total Deaths",
    text="Total Deaths",
    title="Earthquakes Have Caused the Highest Number of Recorded Deaths Worldwide"
)

fig.update_traces(
    texttemplate="%{text:,.0f}",
    textposition="outside"
)

fig.update_layout(
    template="plotly_white",
    title={
        "text":"Earthquakes Have Caused the Highest Number of Recorded Deaths Worldwide",
        "x":0.5,
        "font":{"size":22}
    },
    xaxis_title="Total Deaths",
    yaxis_title="Disaster Type",
    showlegend=False,
    xaxis=dict(showgrid=True, gridcolor="lightgray"),
    yaxis=dict(showgrid=False),
    height=650
)

fig.show()
```

## Interpretation

The chart compares disaster types based on their cumulative recorded fatalities. It highlights which hazards have had the greatest human impact over the study period. These findings help identify disaster categories that require stronger preparedness, early warning systems, and emergency response planning.

# Analytical Question 5

## Is there a relationship between disaster magnitude and the number of deaths?

### Objective

This analysis investigates whether disasters with higher recorded magnitudes generally result in more fatalities.

### Why is this important?

Understanding the relationship between disaster magnitude and fatalities helps evaluate disaster severity and supports improvements in early warning systems, emergency planning, and risk reduction strategies.

```python
q5_df = df[
    [
        "Disaster Type",
        "Dis Mag Value",
        "Total Deaths"
    ]
].copy()

q5_df.head()
```

```python
q5_df.isnull().sum()
```

While larger magnitude disasters often result in greater loss of life, the relationship is not perfectly linear. Some lower-magnitude events have caused significant fatalities due to factors such as population density, infrastructure quality, and preparedness. This indicates that disaster impact depends on both physical intensity and societal vulnerability.

# Analytical Question 6

## Do different disaster types occur more frequently during certain months of the year?

### Objective

This analysis investigates whether natural disasters exhibit seasonal patterns by comparing the monthly occurrence of different disaster types.

### Why is this important?

Identifying seasonal trends helps governments and emergency management agencies prepare resources and develop early warning systems before periods of increased disaster risk.

```python
q6_df = df[
    [
        "Start Month",
        "Disaster Type"
    ]
].copy()

q6_df.head()
```

```python
q6_df.isnull().sum()
```

```python
q6_df = q6_df.dropna(subset=["Start Month"])

q6_df["Start Month"] = q6_df["Start Month"].astype(int)

q6_df.shape
```

```python
heatmap_data = (
    q6_df
    .groupby(["Disaster Type", "Start Month"])
    .size()
    .reset_index(name="Count")
)

heatmap_data = heatmap_data.pivot(
    index="Disaster Type",
    columns="Start Month",
    values="Count"
).fillna(0)

heatmap_data
```

```python
fig = px.imshow(
    heatmap_data,
    text_auto=True,
    aspect="auto",
    color_continuous_scale="Blues",
    title="Seasonal Patterns of Natural Disaster Types"
)

fig.update_layout(
    template="plotly_white",
    title={
        "text": "Seasonal Patterns of Natural Disaster Types",
        "x": 0.5,
        "font": {"size": 22}
    },
    xaxis_title="Month",
    yaxis_title="Disaster Type",
    height=700
)

fig.show()
```

## Interpretation

The heatmap reveals seasonal variations in the occurrence of different disaster types. Darker cells indicate months with higher frequencies of specific disasters, highlighting recurring seasonal patterns. These insights can support seasonal preparedness, resource allocation, and disaster risk management by identifying periods of elevated risk for particular hazards.

# Analytical Question 7

## Which disaster subgroups contribute the most to different disaster types?

### Objective

This analysis examines how disaster events are distributed across disaster subgroups and disaster types between 1970 and 2021.

### Why is this important?

Understanding the composition of disaster events provides a deeper understanding of global disaster patterns and highlights which subgroups dominate different categories of natural disasters.

```python
q7_df = df[
    [
        "Disaster Subgroup",
        "Disaster Type"
    ]
].copy()

q7_df.head()
```

```python
q7_df.isnull().sum()
```

```python
q7_df = q7_df.dropna()

q7_df.shape
```

```python
treemap_data = (
    q7_df
    .groupby(
        ["Disaster Subgroup", "Disaster Type"]
    )
    .size()
    .reset_index(name="Number of Disasters")
)

treemap_data.head()
```

```python
fig = px.treemap(
    treemap_data,
    path=["Disaster Subgroup", "Disaster Type"],
    values="Number of Disasters",
    color="Number of Disasters",
    color_continuous_scale="Blues",
    title="Distribution of Natural Disaster Events by Subgroup and Type"
)

fig.update_layout(
    template="plotly_white",
    title={
        "text": "Distribution of Natural Disaster Events by Subgroup and Type",
        "x": 0.5,
        "font": {"size": 22}
    },
    height=750,
    margin=dict(t=60, l=20, r=20, b=20)
)

fig.show()
```

## Interpretation

The treemap illustrates the hierarchical distribution of disaster events across disaster subgroups and disaster types. Larger rectangles represent disaster categories with a higher number of recorded events. This visualization highlights which subgroups dominate global disaster occurrences and reveals the relative contribution of each disaster type within its subgroup.

# Analytical Question 8

## How do disaster types compare in terms of human impact and economic losses?

### Objective

This analysis compares disaster types based on the total number of people affected and the total economic losses they caused.

### Why is this important?

Some disaster types primarily affect people, while others result in greater financial losses. Comparing these impacts helps identify disaster types that require different response and mitigation strategies.

```python
q8_df = df[
    [
        "Disaster Type",
        "Total Affected",
        "Total Damages ('000 US$)"
    ]
].copy()

q8_df.head()
```

```python
q8_df.isnull().sum()
```

```python
q8_df["Total Affected"] = q8_df["Total Affected"].fillna(0)

q8_df["Total Damages ('000 US$)"] = (
    q8_df["Total Damages ('000 US$)"].fillna(0)
)
```

```python
bubble_data = (
    q8_df
    .groupby("Disaster Type", as_index=False)
    .agg({
        "Total Affected": "sum",
        "Total Damages ('000 US$)": "sum"
    })
)

bubble_data
```

```python
fig = px.scatter(
    bubble_data,
    x="Total Affected",
    y="Total Damages ('000 US$)",
    size="Total Damages ('000 US$)",
    color="Disaster Type",
    hover_name="Disaster Type",
    size_max=60,
    title="Human Impact versus Economic Losses by Disaster Type",
    labels={
        "Total Affected": "Total People Affected",
        "Total Damages ('000 US$)": "Total Economic Damage ('000 US$)"
    }
)

fig.update_layout(
    template="plotly_white",
    title={
        "text": "Human Impact versus Economic Losses by Disaster Type",
        "x": 0.5,
        "font": {"size": 22}
    },
    xaxis=dict(showgrid=False),
    yaxis=dict(showgrid=True, gridcolor="lightgray"),
    height=700
)

fig.show()
```

## Interpretation

The bubble chart compares disaster types based on their cumulative human and economic impacts. Disaster types located in the upper-right corner have both high economic losses and a large number of affected people, while those positioned elsewhere indicate different impact profiles. This comparison highlights that disaster severity cannot be assessed using a single measure and that both humanitarian and economic consequences should be considered.

# Analytical Question 9

## Where are the global hotspots of recorded natural disasters?

### Objective

This analysis visualizes the geographic distribution of recorded natural disasters using their latitude and longitude coordinates.

### Why is this important?

Understanding the spatial distribution of disasters helps identify geographic hotspots, supports disaster risk assessment, and assists policymakers in prioritizing regions for preparedness and resilience planning.

```python
q9_df = df[
    [
        "Latitude",
        "Longitude",
        "Disaster Type"
    ]
].copy()

q9_df.head()
```

```python
q9_df.isnull().sum()
```

```python
q9_df = q9_df.dropna(
    subset=[
        "Latitude",
        "Longitude"
    ]
)

q9_df.shape
```

```python
q9_df["Latitude"] = pd.to_numeric(
    q9_df["Latitude"],
    errors="coerce"
)

q9_df["Longitude"] = pd.to_numeric(
    q9_df["Longitude"],
    errors="coerce"
)

q9_df = q9_df.dropna(
    subset=[
        "Latitude",
        "Longitude"
    ]
)
```

```python
fig = px.scatter_geo(
    q9_df,
    lat="Latitude",
    lon="Longitude",
    color="Disaster Type",
    hover_name="Disaster Type",
    opacity=0.7,
    title="Global Distribution of Recorded Natural Disasters (1970–2021)"
)

fig.update_layout(
    template="plotly_white",
    title={
        "text": "Global Distribution of Recorded Natural Disasters (1970–2021)",
        "x": 0.5,
        "font": {"size": 22}
    },
    height=700,
    geo=dict(
        showland=True,
        landcolor="rgb(240,240,240)",
        showcountries=True,
        countrycolor="gray",
        showocean=True,
        oceancolor="lightblue"
    )
)

fig.show()
```

## Interpretation

The geographic map highlights the spatial distribution of recorded natural disasters worldwide between 1970 and 2021. Clusters of disaster events reveal regions with higher exposure to natural hazards. The visualization demonstrates that disaster occurrence is unevenly distributed across the globe, emphasizing the importance of location-specific disaster preparedness and risk management strategies.

# Analytical Question 10

## How are economic losses distributed across continents and disaster types?

### Objective

This analysis explores how the total economic losses from natural disasters are distributed across continents and disaster types.

### Why is this important?

Different continents experience different disaster profiles and economic consequences. Understanding these patterns helps policymakers prioritize investments in disaster resilience and infrastructure.

```python
q10_df = df[
    [
        "Continent",
        "Disaster Type",
        "Total Damages ('000 US$)"
    ]
].copy()

q10_df.head()
```

```python
q10_df.isnull().sum()
```

```python
q10_df = q10_df.dropna(
    subset=[
        "Continent",
        "Disaster Type"
    ]
)

q10_df["Total Damages ('000 US$)"] = (
    q10_df["Total Damages ('000 US$)"].fillna(0)
)
```

```python
sunburst_data = (
    q10_df
    .groupby(
        ["Continent", "Disaster Type"],
        as_index=False
    )
    ["Total Damages ('000 US$)"]
    .sum()
)

sunburst_data.head()
```

```python
fig = px.sunburst(
    sunburst_data,
    path=["Continent", "Disaster Type"],
    values="Total Damages ('000 US$)",
    color="Total Damages ('000 US$)",
    color_continuous_scale="Blues",
    title="Economic Losses by Continent and Disaster Type"
)

fig.update_layout(
    template="plotly_white",
    title={
        "text": "Economic Losses by Continent and Disaster Type",
        "x": 0.5,
        "font": {"size": 22}
    },
    height=750,
    margin=dict(t=60, l=20, r=20, b=20)
)

fig.show()
```



## Interpretation

The sunburst chart illustrates how economic losses are distributed across continents and disaster types. The size of each segment represents the cumulative financial impact, allowing comparisons between regions and disaster categories. The visualization reveals which combinations contribute the most to global disaster-related economic losses and highlights regional differences in disaster vulnerability.
